```python
# pipeline function that adds arena info and time zone
import pandas as pd
from datetime import datetime, time
from zoneinfo import ZoneInfo


def process_season(filename, season, arena_df):
    # 1. Load raw data
    df_rest = pd.read_csv(filename)

    # 2. Keep the columns used in our analysis
    columns_to_keep = [
        "SEASON_ID",
        "TEAM_ID",
        "TEAM_ABBREVIATION",
        "TEAM_NAME",
        "GAME_ID",
        "GAME_DATE",
        "MATCHUP",
        "WL",
        "PTS",
        "PLUS_MINUS",
        "VIDEO_AVAILABLE",
    ]
    df_rest = df_rest[columns_to_keep].copy()

    # 3. Add the team's geographic region
    df_rest["REGION"] = df_rest["TEAM_ABBREVIATION"].map(region_mapping)

    # 4. Convert dates and sort games chronologically
    df_rest["GAME_DATE"] = pd.to_datetime(df_rest["GAME_DATE"])

    df_rest = df_rest.sort_values(
        ["TEAM_ID", "GAME_DATE", "GAME_ID"]
    ).reset_index(drop=True)

    # 5. Calculate rest days between consecutive games
    df_rest["REST_DAYS"] = (
        df_rest.groupby("TEAM_ID")["GAME_DATE"]
        .diff()
        .dt.days - 1
    )

    # 6. Identify home and away games
    df_rest["LOCATION_TYPE"] = df_rest["MATCHUP"].apply(
        lambda x: "Home" if "vs." in x else "Away"
    )

    # 7. Identify the opponent
    df_rest["OPPONENT_ABBREVIATION"] = (
        df_rest["MATCHUP"].str[-3:]
    )

    # 8. Identify the team hosting the game
    df_rest["GAME_VENUE_TEAM"] = df_rest.apply(
        lambda row: (
            row["TEAM_ABBREVIATION"]
            if row["LOCATION_TYPE"] == "Home"
            else row["OPPONENT_ABBREVIATION"]
        ),
        axis=1
    )

    # 9. Prepare arena information for merging
    arena_lookup = arena_df[
        [
            "TEAM_ABBREVIATION",
            "ARENA_STATE",
            "TIME_ZONE",
            "LATITUDE",
            "LONGITUDE",
        ]
    ].rename(
        columns={
            "TEAM_ABBREVIATION": "GAME_VENUE_TEAM",
            "LATITUDE": "VENUE_LATITUDE",
            "LONGITUDE": "VENUE_LONGITUDE",
        }
    )

    # 10. Add the game venue's state, time zone, and coordinates
    df_rest = df_rest.merge(
        arena_lookup,
        on="GAME_VENUE_TEAM",
        how="left",
        validate="many_to_one"
    )

    # 11. Calculate the venue's UTC offset on each game date
    def get_utc_offset(date, timezone):
        if pd.isna(date) or pd.isna(timezone):
            return pd.NA

        local_time = datetime.combine(
            date.date(),
            time(12, 0),
            tzinfo=ZoneInfo(timezone)
        )

        return local_time.utcoffset().total_seconds() / 3600

    df_rest["UTC_OFFSET_HOURS"] = df_rest.apply(
        lambda row: get_utc_offset(
            row["GAME_DATE"], row["TIME_ZONE"]
        ),
        axis=1
    )

    # 12. Add the season label
    df_rest["SEASON"] = season

    # 13. Check for missing arena information
    check_columns = [
        "ARENA_STATE",
        "TIME_ZONE",
        "VENUE_LATITUDE",
        "VENUE_LONGITUDE",
    ]

    print(f"\nSeason: {season}")
    print(f"Rows: {len(df_rest)}")
    print(f"Unique games: {df_rest['GAME_ID'].nunique()}")
    print("Missing arena information:")
    print(df_rest[check_columns].isna().sum())

    # 14. Save the processed data
    output_filename = f"df_rest_{season.replace('-', '_')}.csv"
    df_rest.to_csv(output_filename, index=False)

    print(f"Saved: {output_filename}")

    return df_rest
```

```python
# calculating distance over earth surface


import numpy as np
import pandas as pd

# function of math method
def haversine_miles(lat1, lon1, lat2, lon2):
    """Calculate great-circle distance between two coordinates, in miles."""

    lat1, lon1, lat2, lon2 = np.radians(
        [lat1, lon1, lat2, lon2]
    )

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1) * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    # Protect against tiny floating-point errors
    a = np.clip(a, 0, 1)

    earth_radius_miles = 3958.8

    return 2 * earth_radius_miles * np.arcsin(np.sqrt(a))

# function to calculate distance 


def add_travel_distance(df_rest, arena_df):
    df = df_rest.copy()

    # Ensure games are in chronological order for each team
    df = df.sort_values(
        ["TEAM_ID", "GAME_DATE", "GAME_ID"]
    ).reset_index(drop=True)

    # Coordinates of the previous game venue
    df["PREV_VENUE_LATITUDE"] = (
        df.groupby("TEAM_ID")["VENUE_LATITUDE"].shift(1)
    )

    df["PREV_VENUE_LONGITUDE"] = (
        df.groupby("TEAM_ID")["VENUE_LONGITUDE"].shift(1)
    )

    # Build a lookup of each team's home arena coordinates
    home_arenas = arena_df.set_index("TEAM_ABBREVIATION")

    # Identify each team's first game within this season
    first_game = df.groupby("TEAM_ID").cumcount() == 0

    # For first games, assume the team starts at its home arena
    df.loc[first_game, "PREV_VENUE_LATITUDE"] = (
        df.loc[first_game, "TEAM_ABBREVIATION"]
        .map(home_arenas["LATITUDE"])
    )

    df.loc[first_game, "PREV_VENUE_LONGITUDE"] = (
        df.loc[first_game, "TEAM_ABBREVIATION"]
        .map(home_arenas["LONGITUDE"])
    )

    # Calculate distance from previous venue to current venue
    df["TRAVEL_DISTANCE_MILES"] = haversine_miles(
        df["PREV_VENUE_LATITUDE"].to_numpy(),
        df["PREV_VENUE_LONGITUDE"].to_numpy(),
        df["VENUE_LATITUDE"].to_numpy(),
        df["VENUE_LONGITUDE"].to_numpy()
    )

    return df

# call functions


df_rest_2324 = add_travel_distance(
    df_rest_2324, arena_df_2324
)

df_rest_2425 = add_travel_distance(
    df_rest_2425, arena_df_2425
)

df_rest_2526 = add_travel_distance(
    df_rest_2526, arena_df
)

```
