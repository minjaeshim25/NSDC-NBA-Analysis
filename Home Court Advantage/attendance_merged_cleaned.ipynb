```python
import pandas as pd
data1 = pd.read_csv("BR.txt", sep = ",")

data2 = pd.read_csv("Team Summaries.csv")

left = data1
right = data2

df = pd.merge(data1, data2, left_on=['Franchise', 'Lg'], 
    right_on=['team', 'lg'])

df

df1 = df[df ['To'] >= '2014-2015']
df2 = df1[df1 ['season'] >= 2014]

df3 = df2.drop(columns = ['abbreviation', 'playoffs', 'age', 'w', 'l', 'pw', 'pl', 'mov', 'sos', 'srs', 'o_rtg', 'd_rtg', 'n_rtg', 'pace', 
                         'f_tr', 'x3p_ar', 'ts_percent', 'e_fg_percent', 'tov_percent', 'orb_percent', 'ft_fga', 'opp_e_fg_percent', 
                         'opp_tov_percent', 'drb_percent', 'opp_ft_fga','lg', 'G','W/L%', 'W', 'L', 'Div', 'Conf', 'Champ', 'Plyfs', 'To', 'From', 'Yrs', 'team'])



df_cleaned = df3.drop_duplicates()

df_cleaned

df_cleaned.to_csv('Arena_Attendance_per_Season.csv', index=False)

# source: https://pandas.pydata.org/docs/user_guide/merging.html
```
