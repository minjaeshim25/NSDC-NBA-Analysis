```python
import pandas as pd

FILE_0304 = "filtered3pt2003-2004 - Sheet1.csv"
FILE_0405 = "filtered3pt2004-2005 - Sheet1.csv"

def load(path):
    d = pd.read_csv(path)
    d.columns = [str(c).strip() for c in d.columns]
    d['clean_name'] = d['Player'].astype(str).str.strip()
    d = d[~d['clean_name'].isin(['nan', ''])]
    return d.drop_duplicates('clean_name')

a, b = load(FILE_0304), load(FILE_0405)
m = pd.merge(a, b, on='clean_name', suffixes=('_0304', '_0405'))

tov_ppg = m[['clean_name', 'PTS_0304', 'PTS_0405', 'TOV_0304', 'TOV_0405']].copy()
tov_ppg.columns = ['Player', 'PPG_0304', 'PPG_0405', 'TOV_0304', 'TOV_0405']
tov_ppg['PPG_change'] = (tov_ppg['PPG_0405'] - tov_ppg['PPG_0304']).round(1)
tov_ppg['TOV_change'] = (tov_ppg['TOV_0405'] - tov_ppg['TOV_0304']).round(1)

print(f"{len(tov_ppg)} players in BOTH files")
print("Missing values:", int(tov_ppg.isna().sum().sum()))
print("\nAverages:")
print(tov_ppg[['PPG_0304', 'PPG_0405', 'TOV_0304', 'TOV_0405']].mean().round(2))
tov_ppg
```
