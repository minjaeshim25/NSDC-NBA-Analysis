```python
import pandas as pd

# 1. Load the raw files fresh
file_04 = "filteredmid2003-2004 - Sheet1.csv"
file_05 = "filteredmid2004-2005 - Sheet1.csv"

# header=None lets us pick the columns by index number
df_04_raw = pd.read_csv(file_04, header=None)
df_05_raw = pd.read_csv(file_05, header=None)

# 2. Skip the first 4 rows of messy headers to get straight to the raw numbers
# Row 0 is the player name, column 11 is L, column 12 is M, column 13 is N
df_04_clean = df_04_raw.iloc[4:].copy()
df_05_clean = df_05_raw.iloc[4:].copy()

# 3. Create a clean dataframe for 2003-04
data_0304 = pd.DataFrame()
data_0304['clean_name'] = df_04_clean[0].astype(str).str.strip()
data_0304['MidRange_FGM_0304'] = pd.to_numeric(df_04_clean[11], errors='coerce')
data_0304['MidRange_FGA_0304'] = pd.to_numeric(df_04_clean[12], errors='coerce')
data_0304['MidRange_FG%_0304'] = pd.to_numeric(df_04_clean[13], errors='coerce')

# 4. Create a clean dataframe for 2004-05
data_0405 = pd.DataFrame()
data_0405['clean_name'] = df_05_clean[0].astype(str).str.strip()
data_0405['MidRange_FGM_0405'] = pd.to_numeric(df_05_clean[11], errors='coerce')
data_0405['MidRange_FGA_0405'] = pd.to_numeric(df_05_clean[12], errors='coerce')
data_0405['MidRange_FG%_0405'] = pd.to_numeric(df_05_clean[13], errors='coerce')

# 5. Clean up duplicates to avoid any data multipliers
data_0304 = data_0304.drop_duplicates(subset=['clean_name'])
data_0405 = data_0405.drop_duplicates(subset=['clean_name'])

# 6. Cross-reference inner merge to match the players
matched_df = pd.merge(data_0304, data_0405, on='clean_name')
matched_df.rename(columns={'clean_name': 'Player'}, inplace=True)

# 7. Arrange the layout into 3 separate side-by-side blocks
final_columns = [
    'Player',
    'MidRange_FGM_0304', 'MidRange_FGM_0405',  # Made side-by-side (L)
    'MidRange_FGA_0304', 'MidRange_FGA_0405',  # Attempts side-by-side (M)
    'MidRange_FG%_0304', 'MidRange_FG%_0405'   # Percentage side-by-side (N)
]

comparison_df = matched_df[final_columns]
comparison_df


```
